```python
import sympy

g, m = sympy.symbols('g m')

t = sympy.Symbol('t')
vx = sympy.Function('v_x')(t)
vy = sympy.Function('v_y')(t)
vz = sympy.Function('v_z')(t)

phi_d = sympy.atan2(-vy, vz + g)
theta_d = sympy.atan2(vx*sympy.cos(phi_d), vz + g)
F_ref = m*(vz + g)/(sympy.cos(phi_d)*sympy.cos(theta_d))

from IPython.display import display, Math

subs = [
    (sympy.Derivative(vx, t, 2), sympy.Symbol('v2(0)')),
    (sympy.Derivative(vy, t, 2), sympy.Symbol('v2(1)')),
    (sympy.Derivative(vz, t, 2), sympy.Symbol('v2(2)')),
    (sympy.Derivative(vx, t), sympy.Symbol('v1(0)')),
    (sympy.Derivative(vy, t), sympy.Symbol('v1(1)')),
    (sympy.Derivative(vz, t), sympy.Symbol('v1(2)')),
    (vx, sympy.Symbol('v(0)')),
    (vy, sympy.Symbol('v(1)')),
    (vz, sympy.Symbol('v(2)')),
    (m, sympy.Symbol('Params::m')),
    (g, sympy.Symbol('Params::g')),
]

display(Math('F_d = ' + sympy.latex(sympy.simplify(F_ref))))

print(sympy.ccode(sympy.simplify(F_ref).subs(subs)))

display(Math('\phi_d        = ' + sympy.latex(sympy.simplify(phi_d))))
display(Math('\dot{\phi}_d  = ' + sympy.latex(sympy.simplify(phi_d.diff(t)))))
display(Math('\ddot{\phi}_d = ' + sympy.latex(sympy.simplify(phi_d.diff(t, 2)))))

print(sympy.ccode(sympy.simplify(phi_d.subs(subs))))
print(sympy.ccode(sympy.simplify(phi_d.diff(t).subs(subs))))
print(sympy.ccode(sympy.simplify(phi_d.diff(t, 2).subs(subs))))

display(Math('\\theta_d        = ' + sympy.latex(sympy.simplify(theta_d))))
display(Math('\dot{\\theta}_d  = ' + sympy.latex(sympy.simplify(theta_d.diff(t)))))
display(Math('\ddot{\\theta}_d = ' + sympy.latex(sympy.simplify(theta_d.diff(t, 2)))))

print(sympy.ccode(sympy.simplify(theta_d.subs(subs))))
print(sympy.ccode(sympy.simplify(theta_d.diff(t).subs(subs))))
print(sympy.ccode(sympy.simplify(theta_d.diff(t, 2).subs(subs))))
```
